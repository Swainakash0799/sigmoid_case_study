## Data Analysis, EDA & Hypothesis Testing

Load the dataset



Check data quality

Perform EDA



Create features



Test the working-day hypothesis



Save the cleaned dataset for Stage 2

```python

# ============================================================
# SIGMOID BIKE-SHARING CASE STUDY
# STAGE 1: DATA ANALYSIS, EDA & HYPOTHESIS TESTING
# ============================================================


# STEP 1: IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import ttest_ind


# STEP 2: LOAD DATASET
# ------------------------------------------------------------

# Upload hour.csv in Google Colab
from google.colab import files

uploaded = files.upload()

# Read the uploaded CSV file
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")

print("Dataset shape:", df.shape)

display(df.head())


# STEP 3: UNDERSTAND THE DATA
# ------------------------------------------------------------

# Display column names
print("Column names:")
print(df.columns.tolist())

# Check data types and non-null values
df.info()

# Statistical summary
display(df.describe())


# STEP 4: DATA QUALITY CHECKS
# ------------------------------------------------------------

# Check missing values
print("Missing values:")
print(df.isnull().sum())

print(
    "Total missing values:",
    df.isnull().sum().sum()
)

# Remove rows containing missing values
df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum().sum())

# Check duplicate records
print(
    "Duplicate rows:",
    df.duplicated().sum()
)

# Check duplicate date-hour combinations
print(
    "Duplicate date-hour records:",
    df.duplicated(subset=["dteday", "hr"]).sum()
)

# Check invalid rental counts
print(
    "Negative rental counts:",
    (df["cnt"] < 0).sum()
)

# Check invalid hour values
print(
    "Invalid hour values:",
    (~df["hr"].between(0, 23)).sum()
)


# STEP 5: CLEAN THE DATA
# ------------------------------------------------------------

# Remove duplicate rows
df = df.drop_duplicates()

# Convert date column into datetime format
df["dteday"] = pd.to_datetime(df["dteday"])

# Sort data by date and hour
df = df.sort_values(
    by=["dteday", "hr"]
).reset_index(drop=True)

print("Cleaned dataset shape:", df.shape)


# STEP 6: EXPLORATORY DATA ANALYSIS
# ------------------------------------------------------------

# A. Distribution of bike demand

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="cnt",
    bins=30,
    kde=True
)

plt.title("Distribution of Hourly Bike Demand")
plt.xlabel("Total Bike Rentals")
plt.ylabel("Frequency")

plt.show()


# B. Average demand by hour

hourly_demand = df.groupby("hr")["cnt"].mean()

plt.figure(figsize=(10, 5))

hourly_demand.plot(
    kind="line",
    marker="o"
)

plt.title("Average Bike Demand by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Rentals")

plt.xticks(range(24))
plt.grid()

plt.show()


# C. Average demand by weekday

weekday_demand = df.groupby("weekday")["cnt"].mean()

plt.figure(figsize=(8, 5))

weekday_demand.plot(kind="bar")

plt.title("Average Bike Demand by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Average Rentals")

plt.show()


# D. Demand by weather

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="weathersit",
    y="cnt"
)

plt.title("Bike Demand by Weather")
plt.xlabel("Weather Situation")
plt.ylabel("Total Rentals")

plt.show()


# E. Correlation heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Variables")

plt.show()


# STEP 7: FEATURE ENGINEERING
# ------------------------------------------------------------

# Extract day from date
df["day"] = df["dteday"].dt.day

# Create weekend feature
df["is_weekend"] = df["weekday"].isin([0, 6]).astype(int)

# Create cyclical hour features
df["hour_sin"] = np.sin(
    2 * np.pi * df["hr"] / 24
)

df["hour_cos"] = np.cos(
    2 * np.pi * df["hr"] / 24
)

# Create cyclical month features
df["month_sin"] = np.sin(
    2 * np.pi * df["mnth"] / 12
)

df["month_cos"] = np.cos(
    2 * np.pi * df["mnth"] / 12
)

display(df.head())


# STEP 8: HYPOTHESIS TESTING
# ------------------------------------------------------------

# Hypothesis:
# Average bike demand differs between working days
# and non-working days.

# Null Hypothesis:
# There is no significant difference in average demand.

# Alternative Hypothesis:
# There is a significant difference in average demand.


# Separate demand into two groups

working_day_demand = df[
    df["workingday"] == 1
]["cnt"]

non_working_day_demand = df[
    df["workingday"] == 0
]["cnt"]


# Calculate average demand

working_average = working_day_demand.mean()

non_working_average = non_working_day_demand.mean()


# Perform Welch's t-test

t_stat, p_value = ttest_ind(
    working_day_demand,
    non_working_day_demand,
    equal_var=False
)


# Display results

print("Working day average:", round(working_average, 2))

print(
    "Non-working day average:",
    round(non_working_average, 2)
)

print("T-statistic:", round(t_stat, 4))

print("P-value:", p_value)


# Interpret the result

if p_value < 0.05:

    print("\nReject the null hypothesis.")

    print(
        "There is evidence of a significant "
        "difference in average demand."
    )

else:

    print("\nFail to reject the null hypothesis.")

    print(
        "There is insufficient evidence of "
        "a significant difference."
    )


# STEP 9: SAVE CLEANED DATASET
# ------------------------------------------------------------

df.to_csv(
    "bike_cleaned.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
```

##Machine Learning, Model Comparison & Final Results

Load the cleaned dataset



Prepare input features and target



Split data chronologically



Train Linear Regression and XGBoost



Compare both models



Select the best model



Evaluate final test performance



Generate actual vs predicted results



Create a confusion matrix



Check feature importance



Save the final model

```python

# ============================================================
# SIGMOID BIKE-SHARING CASE STUDY
# STAGE 2: MACHINE LEARNING & FINAL EVALUATION
# ============================================================


# STEP 1: IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import joblib

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report
)

from xgboost import XGBRegressor


# STEP 2: LOAD CLEANED DATASET
# ------------------------------------------------------------

# Upload bike_cleaned.csv in Google Colab
from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

# Convert date column
df["dteday"] = pd.to_datetime(df["dteday"])

# Sort chronologically
df = df.sort_values(
    by=["dteday", "hr"]
).reset_index(drop=True)

print("Dataset loaded successfully!")

print("Dataset shape:", df.shape)

display(df.head())


# STEP 3: PREPARE INPUT FEATURES AND TARGET
# ------------------------------------------------------------

# Features used by the model
features = [
    "season",
    "yr",
    "mnth",
    "hr",
    "weekday",
    "holiday",
    "workingday",
    "weathersit",
    "temp",
    "atemp",
    "hum",
    "windspeed",
    "day",
    "is_weekend",
    "hour_sin",
    "hour_cos",
    "month_sin",
    "month_cos"
]

# Input variables
X = df[features]

# Target variable
y = df["cnt"]

print("Input shape:", X.shape)

print("Target shape:", y.shape)


# STEP 4: SPLIT DATA CHRONOLOGICALLY
# ------------------------------------------------------------

# Total number of records
n = len(df)

# 70% training data
train_end = int(n * 0.70)

# 85% training + validation data
val_end = int(n * 0.85)


# Training data
X_train = X.iloc[:train_end]

y_train = y.iloc[:train_end]


# Validation data
X_val = X.iloc[train_end:val_end]

y_val = y.iloc[train_end:val_end]


# Testing data
X_test = X.iloc[val_end:]

y_test = y.iloc[val_end:]


print("Training records:", len(X_train))

print("Validation records:", len(X_val))

print("Testing records:", len(X_test))


# STEP 5: TRAIN LINEAR REGRESSION
# ------------------------------------------------------------

# Create model
lr_model = LinearRegression()

# Train model
lr_model.fit(X_train, y_train)

# Predict validation demand
lr_predictions = lr_model.predict(X_val)

print("Linear Regression training completed!")


# STEP 6: TRAIN XGBOOST
# ------------------------------------------------------------

# Create XGBoost model
xgb_model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.1,
    max_depth=5,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

# Train model
xgb_model.fit(X_train, y_train)

# Predict validation demand
xgb_predictions = xgb_model.predict(X_val)

print("XGBoost training completed!")


# STEP 7: COMPARE BOTH MODELS
# ------------------------------------------------------------

# Calculate Linear Regression metrics

lr_mae = mean_absolute_error(
    y_val,
    lr_predictions
)

lr_rmse = np.sqrt(
    mean_squared_error(
        y_val,
        lr_predictions
    )
)

lr_r2 = r2_score(
    y_val,
    lr_predictions
)


# Calculate XGBoost metrics

xgb_mae = mean_absolute_error(
    y_val,
    xgb_predictions
)

xgb_rmse = np.sqrt(
    mean_squared_error(
        y_val,
        xgb_predictions
    )
)

xgb_r2 = r2_score(
    y_val,
    xgb_predictions
)


# Create comparison table

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "XGBoost"
    ],

    "MAE": [
        round(lr_mae, 2),
        round(xgb_mae, 2)
    ],

    "RMSE": [
        round(lr_rmse, 2),
        round(xgb_rmse, 2)
    ],

    "R2": [
        round(lr_r2, 4),
        round(xgb_r2, 4)
    ]
})


display(results)


# STEP 8: VISUALIZE MODEL COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=results,
    x="Model",
    y="RMSE"
)

plt.title("Model Comparison Based on RMSE")

plt.xlabel("Model")

plt.ylabel("Validation RMSE")

plt.show()


# STEP 9: SELECT THE BEST MODEL
# ------------------------------------------------------------

# Select model with lowest validation RMSE

if xgb_rmse < lr_rmse:

    best_model_name = "XGBoost"

    print("XGBoost is the better model.")

else:

    best_model_name = "Linear Regression"

    print("Linear Regression is the better model.")


# STEP 10: TRAIN FINAL MODEL
# ------------------------------------------------------------

# Combine training and validation data
# because model selection is complete

X_train_final = pd.concat(
    [X_train, X_val]
)

y_train_final = pd.concat(
    [y_train, y_val]
)


# Train the selected model on combined data

if best_model_name == "XGBoost":

    best_model = XGBRegressor(
        n_estimators=500,
        learning_rate=0.1,
        max_depth=5,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )

else:

    best_model = LinearRegression()


best_model.fit(
    X_train_final,
    y_train_final
)


# STEP 11: FINAL TEST EVALUATION
# ------------------------------------------------------------

# Predict test demand

y_pred = best_model.predict(X_test)


# Calculate final metrics

final_mae = mean_absolute_error(
    y_test,
    y_pred
)

final_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

final_r2 = r2_score(
    y_test,
    y_pred
)


# Display final results

print("\nFINAL MODEL:", best_model_name)

print("------------------------------")

print("Test MAE:", round(final_mae, 2))

print("Test RMSE:", round(final_rmse, 2))

print("Test R2:", round(final_r2, 4))

print(
    "R2 Percentage:",
    round(final_r2 * 100, 2),
    "%"
)


# STEP 12: ACTUAL VS PREDICTED DEMAND
# ------------------------------------------------------------

comparison = pd.DataFrame({

    "Actual Demand": y_test.values,

    "Predicted Demand": np.round(y_pred),

    "Absolute Error": np.round(
        np.abs(y_test.values - y_pred),
        2
    )

})


display(comparison.head(20))


# STEP 13: VISUALIZE ACTUAL VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    y_test.values[:200],
    label="Actual Demand"
)

plt.plot(
    y_pred[:200],
    label="Predicted Demand"
)

plt.title("Actual vs Predicted Bike Demand")

plt.xlabel("Test Records")

plt.ylabel("Bike Rentals")

plt.legend()

plt.show()


# STEP 14: CONFUSION MATRIX
# ------------------------------------------------------------

# Demand categories:
# Low = bottom 25% of training demand
# Medium = middle 50%
# High = top 25%


# Calculate thresholds using training data only

low_threshold = y_train_final.quantile(0.25)

high_threshold = y_train_final.quantile(0.75)


# Function to classify demand

def classify_demand(demand):

    if demand <= low_threshold:

        return "Low"

    elif demand <= high_threshold:

        return "Medium"

    else:

        return "High"


# Convert actual demand into categories

actual_class = pd.Series(
    y_test.values
).apply(classify_demand)


# Convert predicted demand into categories

predicted_class = pd.Series(
    y_pred
).apply(classify_demand)


# Define category order

labels = [
    "Low",
    "Medium",
    "High"
]


# Calculate confusion matrix

cm = confusion_matrix(
    actual_class,
    predicted_class,
    labels=labels
)


# Display confusion matrix

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(cmap="Blues")

plt.title("Bike Demand Confusion Matrix")

plt.show()


# STEP 15: CLASSIFICATION ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(
    actual_class,
    predicted_class
)

print(
    "Demand Category Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


# Classification report

print(
    classification_report(
        actual_class,
        predicted_class,
        labels=labels,
        zero_division=0
    )
)


# STEP 16: FEATURE IMPORTANCE
# ------------------------------------------------------------

if best_model_name == "XGBoost":

    importance = pd.DataFrame({

        "Feature": features,

        "Importance": best_model.feature_importances_

    })


    importance = importance.sort_values(
        by="Importance",
        ascending=False
    )


    display(importance)


    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=importance,
        x="Importance",
        y="Feature"
    )

    plt.title("XGBoost Feature Importance")

    plt.show()


# STEP 17: SAVE FINAL MODEL
# ------------------------------------------------------------

model_data = {

    "model": best_model,

    "features": features,

    "model_name": best_model_name,

    "low_threshold": low_threshold,

    "high_threshold": high_threshold

}


joblib.dump(
    model_data,
    "bike_demand_model.pkl"
)


print("\nFinal model saved successfully!")

print("File name: bike_demand_model.pkl")


# STEP 18: FINAL CONCLUSION
# ------------------------------------------------------------

print("\nPROJECT CONCLUSION")

print("------------------------------")

print(
    "Selected Model:",
    best_model_name
)

print(
    "The model predicts hourly bike demand "
    "using historical time and weather features."
)

print(
    "The results can help the operations team "
    "anticipate periods of high demand."
)
```
